### Boolean Subsetting（数据取子集）

```python
# mydf.isna().sum()                     # 每一列有几个缺失值（最常用）
# mydf['viewCount'].isna()              # 只检查某一列，返回 True / False
# mydf[mydf['viewCount'].isna()]        # 找出 viewCount 是缺失值的行
# mydf[~mydf['viewCount'].isna()]       # 找出 viewCount 不是缺失值的行（~ 表示"相反"）
# 为什么要写两次 mydf？
# 因为两个 mydf 做的是不同的事情。第一步：里面的 mydf，用来"做条件"
# 第二步：外面的 mydf，用来"拿数据"

```

```python
# Boolean subsetting allows us to filter rows based on conditions.
# It's very helpful to filter out rows containing null values.
mydf[mydf.index.isna()]
```

```python
mydf2 = mydf[~mydf.index.isna()]
```

### Casting dtypes

```python
mydf2.dtypes
```

```python
mydf2['viewCount'] = mydf2['viewCount'].astype('int')
```

```python
mydf2.dtypes
```

```python
mydf2['publishTime'] = pd.to_datetime(mydf2['publishTime']) # pd.to_datetime() is a useful method to convert str into datetime format. so similar method like pd.to_numeric()
mydf2.dtypes
```

### Creating New Column

```python
mydf2['likeration'] = mydf2['likeCount'] / mydf2['viewCount']
mydf2.head(5)
```

### Adding New Row

```python
# pd.concat() 把多个 DataFrame 上下拼接在一起（默认按行，axis=0）
# 这里把 mydf2 的最后一行（tail(1)）再接到 mydf2 底部，所以最后一行会重复出现
# 注意：要拼接的 DataFrame 必须放在 list [ ] 里面
# 拼接后 index 会重复（最后两行的 index 一样），想重新编号可以加 ignore_index=True

tail = mydf2.tail(1)
concat_exp = pd.concat([mydf2, tail])
concat_exp
```
